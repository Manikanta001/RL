```python
# Greedy Algorithm using Arithmetic Mean

machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

max_q = 0
selected = ""

for machine in machines:
    rewards = machines[machine]

    q = sum(rewards) / len(rewards)

    print("Machine", machine)
    print("Rewards:", rewards)
    print("Q(" + machine + ") =", q)
    print()

    if q > max_q:
        max_q = q
        selected = machine

print("Greedy Selected Machine:", selected)
```

```python
# Greedy Algorithm using Incremental Update

machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

q = {
    "A": 0,
    "B": 0,
    "C": 0
}

count = {
    "A": 0,
    "B": 0,
    "C": 0
}

for machine in machines:
    print("Machine", machine)

    for reward in machines[machine]:
        count[machine] = count[machine] + 1

        q[machine] = q[machine] + (
            reward - q[machine]
        ) / count[machine]

        print("Reward =", reward,
              "N =", count[machine],
              "Q =", round(q[machine], 2))

    print()

selected = "A"

for machine in q:
    if q[machine] > q[selected]:
        selected = machine

print("Final Q Values:")

for machine in q:
    print("Q(" + machine + ") =", round(q[machine], 2))

print("Greedy Selected Machine:", selected)
```
