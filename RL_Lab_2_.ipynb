```python
# Epsilon-Greedy Strategy

machines = ['A', 'B', 'C']

rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

epsilon = 0.2

# Fixed machine selections for 10 plays
selected = [0, 0, 1, 1, 2, 1, 0, 2, 1, 0]

position = [0, 0, 0]
total = 0

for i in range(10):

    index = selected[i]
    machine = machines[index]

    reward = rewards[machine][position[index]]
    position[index] += 1

    N[index] += 1
    R[index] += reward

    Q[index] = Q[index] + (reward - Q[index]) / N[index]

    total += reward

    print("Play", i + 1,
          "Machine =", machine,
          "Reward =", reward,
          "Q =", round(Q[index], 2))

print("\nFinal Results")

for i in range(3):
    print(machines[i],
          "N =", N[i],
          "R =", R[i],
          "Q =", round(Q[i], 2))

print("\nTotal Reward =", total)
```
